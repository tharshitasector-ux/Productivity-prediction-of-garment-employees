```python
target_column = "actual_productivity"

# Separate input features (X) and target (y).
X = df.drop(columns=[target_column])
y = df[target_column]

# Identify categorical and numerical columns.
categorical_columns = X.select_dtypes(include=["object", "str"]).columns.tolist()
numeric_columns = X.select_dtypes(include=["number"]).columns.tolist()

# Display the results.
print("Categorical columns:", categorical_columns)
print("Numerical columns:", numeric_columns)
print("Independent variables (X):", X.shape)
print("Dependent variable (y):", y.shape)
```

```python
# Select numerical input columns only (do not include the target here).
outlier_columns = X.select_dtypes(include="number").columns.tolist()

plt.figure(figsize=(14, 6))
sns.boxplot(data=X[outlier_columns])
plt.title("Boxplot Before Outlier Treatment")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
def cap_outliers_iqr(dataframe, columns):
    dataframe = dataframe.copy()

    for column in columns:
        q1 = dataframe[column].quantile(0.25)
        q3 = dataframe[column].quantile(0.75)
        iqr = q3 - q1

        lower_limit = q1 - 1.5 * iqr
        upper_limit = q3 + 1.5 * iqr

        dataframe[column] = dataframe[column].clip(
            lower=lower_limit,
            upper=upper_limit
        )

    return dataframe

X = cap_outliers_iqr(X, outlier_columns)

plt.figure(figsize=(14, 6))
sns.boxplot(data=X[outlier_columns])
plt.title("Boxplot After Outlier Treatment")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# X contains independent variables; y contains actual productivity.
print("X shape:", X.shape)
print("y shape:", y.shape)

print("\\nTarget variable summary:")
print(y.describe())
```

```python
RANDOM_STATE = 42

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
```

```python
# Identify columns using the training data.
numeric_features = X_train.select_dtypes(include="number").columns.tolist()
categorical_features = X_train.select_dtypes(include=["object", "str"]).columns.tolist()
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_transformer, numeric_features),
        ("categorical", categorical_transformer, categorical_features)
    ],
    remainder="drop"
)

print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)
```

```python
# Fit preprocessing only on the training data before feature selection.
X_train_prepared = preprocessor.fit_transform(X_train)
X_test_prepared = preprocessor.transform(X_test)

feature_names = preprocessor.get_feature_names_out()

# Select up to 10 useful features (or all, if fewer than 10 exist).
k_features = min(10, X_train_prepared.shape[1])

selector = SelectKBest(score_func=f_regression, k=k_features)
X_train_selected = selector.fit_transform(X_train_prepared, y_train)
X_test_selected = selector.transform(X_test_prepared)

selected_features = feature_names[selector.get_support()]

print("Selected Features:")
print(selected_features)
```

```python
feature_scores = pd.DataFrame({
    "Feature": feature_names,
    "Score": selector.scores_,
    "Selected": selector.get_support()
}).sort_values(by="Score", ascending=False)

display(feature_scores)
```

```python
models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=RANDOM_STATE),
    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=RANDOM_STATE
    )
}

trained_models = {}

for name, model in models.items():
    model.fit(X_train_selected, y_train)
    trained_models[name] = model
    print(name, "trained successfully!")
```

```python
results = []

for name, model in trained_models.items():
    y_pred = model.predict(X_test_selected)

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    })

results_df = pd.DataFrame(results).sort_values(by="R2 Score", ascending=False)
display(results_df)
```
